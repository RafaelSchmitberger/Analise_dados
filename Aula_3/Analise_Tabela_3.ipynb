# Análise da Tabela 3 — Áreas de formação na graduação

Dados do IBGE (Censo Demográfico), separados pelas cinco grandes regiões do Brasil.

```python
import pandas as pd
from pathlib import Path

path = Path('.')
arquivo_original = path / 'Tabela_3_Areas_gerais_de_formacao_na_graduacao.pdf'
if not arquivo_original.exists():
    raise FileNotFoundError('Coloque a Tabela 3 na mesma pasta do notebook.')
```
